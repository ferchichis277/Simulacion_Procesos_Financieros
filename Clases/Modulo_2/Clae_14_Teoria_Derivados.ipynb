# Clase 14: Teoría de derivados financieros

Las opciones son instrumentos cuyo valor depende de la evolución futura de otro activo. Debido a que el precio futuro de ese activo es incierto, la simulación se convierte en una herramienta natural para estimar el valor de estos contratos.

Antes de comenzar a valuar opciones, revisaremos algunos conceptos básicos.

# 1. Derivados

Un derivado es un instrumento financiero cuyo valor depende del precio de otro activo, llamado activo subyacente.

El activo subyacente puede ser, por ejemplo:

- una acción
- un índice bursátil
- una divisa
- una materia prima
- una tasa de interés

El precio del derivado se determina a partir del comportamiento del activo subyacente.



Existen varios tipos de derivados. Algunos de los más comunes son:

- Forwards
- Futuros
- Opciones
- Swaps
- Derivados de crédito

---

## Futuros

Los **futuros** son similares a los forwards, pero se negocian en **mercados organizados** (bolsas de derivados).

Estos contratos están **estandarizados** y suelen requerir depósitos de garantía y ajustes diarios de ganancias y pérdidas.

## Forwards

Un **forward** es un contrato entre dos partes para comprar o vender un activo en una fecha futura a un precio acordado hoy.

Estos contratos suelen negociarse **de forma privada (over-the-counter)** entre instituciones financieras o empresas.

---

## Opciones

Una opción es un contrato que otorga a su comprador el derecho, pero no la obligación, de comprar o vender un activo a un precio predeterminado.

Debido a esta característica, las opciones tienen una estructura de payoff distinta a la de forwards y futuros.

En esta materia nos enfocaremos principalmente en la valuación de opciones, ya que presentan varios problemas interesantes desde el punto de vista de la simulación.

---

## Swaps

Un swap es un contrato en el que dos partes acuerdan intercambiar flujos de efectivo en el futuro según ciertas reglas.

Uno de los ejemplos más comunes es el interest rate swap, donde una parte paga una tasa fija y recibe una tasa variable sobre un mismo monto nocional.

Los swaps son ampliamente utilizados por empresas e instituciones financieras para gestionar riesgos de tasas de interés o de tipo de cambio.

---

## Derivados de crédito

Los derivados de crédito son contratos cuyo valor depende del riesgo de incumplimiento (default) de un emisor.

Entre estos instrumentos se encuentran productos estructurados como:

- CDO (Collateralized Debt Obligations)
- otros instrumentos basados en carteras de deuda

Estos derivados permiten transferir o redistribuir el riesgo de crédito entre diferentes inversionistas.

## Propósitos de los derivados

Los derivados cumplen varios roles en los mercados financieros.



### Cobertura (Hedging)

Permiten **reducir o transferir riesgo**.

Por ejemplo, una empresa que recibirá pagos en dólares en el futuro puede utilizar derivados para fijar hoy el tipo de cambio y evitar pérdidas si la moneda se deprecia.

De forma similar, un productor de materias primas puede utilizar derivados para fijar el precio al que venderá su producción en el futuro.

---

### Especulación

Los derivados también permiten **tomar posiciones sobre movimientos futuros de precios**.

Debido a que muchos derivados requieren una inversión inicial relativamente pequeña en comparación con el valor del activo subyacente, pueden amplificar ganancias o pérdidas.

Por esta razón, los derivados son instrumentos comunes entre traders y fondos de inversión.

---

### Arbitraje

Los derivados también se utilizan para **aprovechar inconsistencias de precios entre mercados**.

Si el precio de un derivado no es consistente con el precio del activo subyacente, es posible construir estrategias que generen ganancias sin riesgo utilizando ambos instrumentos.

Gran parte de la teoría moderna de valuación de derivados se basa en la idea de que en ausencia de oportunidades de arbitraje los precios deben seguir ciertas relaciones matemáticas.

---

# 2. Opciones

Una **opción** es un contrato que otorga a su comprador el derecho, pero no la obligación, de comprar o vender un activo a un precio predeterminado en una fecha futura o antes de ella.

Existen dos tipos principales de opciones:

## Call

Una **opción call** otorga el derecho de **comprar** el activo subyacente a un precio predeterminado.

## Put

Una **opción put** otorga el derecho de **vender** el activo subyacente a un precio predeterminado.

# 3. Posiciones en opciones

Cada contrato de opción tiene dos partes: comprador y vendedor.

Las posiciones se pueden resumir así.

### Call

- Comprador de call: tiene el derecho de comprar el activo al precio strike.
- Vendedor de call: tiene la obligación de vender el activo al precio strike si el comprador decide ejercer.

### Put

- Comprador de put: tiene el derecho de vender el activo al precio strike.
- Vendedor de put: tiene la obligación de comprar el activo al precio strike si el comprador decide ejercer.

Solo el comprador tiene la decisión de ejercer o no la opción.

# 4. Prima de la opción

Comprar una opción tiene un costo llamado prima (*premium*).

- El comprador paga la prima.
- El vendedor recibe la prima.

La prima representa el **precio del contrato de opción en el mercado**.

Una pregunta central en finanzas cuantitativas es:

**¿Cuál debería ser el valor justo de esa prima?**

Responder esta pregunta es el objetivo de los **modelos de valuación de opciones**.

# 5. Estilos de opciones

Las opciones pueden clasificarse según **las reglas que determinan cuándo y cómo pueden ejercerse**. Estas reglas afectan tanto el comportamiento del contrato como la dificultad de su valuación.

---

### Opciones europeas

Una opción europea solo puede ejercerse en la fecha de vencimiento.

Esto significa que el valor de la opción depende únicamente del precio del activo subyacente en ese momento. 

En algunos casos incluso existen fórmulas analíticas para calcular su precio. Sin embargo, también pueden valuarse mediante simulación.

En este curso comenzaremos trabajando con opciones europeas, ya que permiten introducir de forma clara los métodos de simulación.

### Opciones americanas

Una opción americana puede ejercerse en cualquier momento antes o en la fecha de vencimiento.

Esto introduce una decisión adicional: el tenedor de la opción debe determinar cuándo es óptimo ejercerla.

Debido a esta característica, la valuación de opciones americanas es más compleja. El valor de la opción depende no solo del precio final del activo, sino también de la posibilidad de ejercer en momentos intermedios.

Para este tipo de opciones se utilizan comúnmente métodos numéricos, como los modelos de árboles binomiales.

En este curso veremos cómo valuar opciones americanas utilizando este tipo de modelos.

### Otras estructuras de opciones

Además de las opciones europeas y americanas, existen muchas otras variantes de opciones cuyo payoff depende de distintas características del comportamiento del activo subyacente.

Algunos ejemplos son:

**Opciones asiáticas**

En las opciones asiáticas, el payoff depende del promedio del precio del activo durante un periodo de tiempo, en lugar de depender únicamente del precio en el vencimiento.

Este tipo de opciones aparece con frecuencia en mercados de materias primas y energía, ya que el uso de promedios reduce el impacto de movimientos extremos en el precio.

**Opciones con barrera**

Las opciones con barrera dependen de si el precio del activo alcanza o cruza ciertos niveles durante la vida del contrato.

Por ejemplo, algunas opciones dejan de existir si el precio supera un nivel específico, mientras que otras solo se activan si ese nivel es alcanzado.

Estas características hacen que su valuación dependa de la trayectoria del precio del activo a lo largo del tiempo, no solo de su valor final.

# 6. Características del contrato

Un contrato de opción típicamente especifica:

- el tipo de opción (call o put)
- el activo subyacente
- la cantidad del activo
- el precio de ejercicio (strike)
- la fecha de vencimiento
- el estilo de ejercicio

---

# 7. Payoff de una opción

El **payoff** es el valor que recibe el tenedor de la opción en el momento del vencimiento.

Para una **call europea**:

$\max(S_T - K,0)$

Para una **put europea**:

$\max(K - S_T,0)$

donde:

- ($S_T$) es el precio del activo al vencimiento
- ($K$) es el precio strike.

---

## Forma del payoff



### Payoff de una call

Si el precio final del activo está por debajo del strike, la opción expira sin valor.
Si está por encima, el payoff aumenta linealmente.

---

### Payoff de una put

Si el precio final del activo está por encima del strike, la opción no se ejerce.
Si está por debajo, el payoff aumenta conforme cae el precio del activo.

---

# 8. Moneyness

La relación entre el precio del activo y el strike determina si una opción está:

- **In the Money (ITM)**

La opción tiene **valor intrínseco positivo**.

Si se ejerciera inmediatamente, el comprador podría adquirir el activo al precio KK y venderlo en el mercado a SS, obteniendo una ganancia de S−KS−K.

- **At the Money (ATM)**

El precio del activo es aproximadamente igual al strike.

Ejercer la opción no genera ganancia ni pérdida inmediata, por lo que el valor intrínseco es cercano a cero.

- **Out of the Money (OTM)**

El precio del activo está por debajo del strike.

Ejercer la opción no tiene sentido económico, ya que sería posible comprar el activo más barato en el mercado.

Para una **call**:

| Situación | Condición |
| --- | --- |
| In the Money | ($S > K$) |
| At the Money | ($S \approx K$) |
| Out of the Money | ($S < K$) |

Para una **put** ocurre lo contrario.

Este concepto es útil porque describe cuándo una opción tiene valor intrínseco.



# 9. Valor intrínseco y valor temporal

El precio de una opción puede descomponerse en dos componentes.

## Valor intrínseco

Es el valor que tendría la opción **si se ejerciera inmediatamente**.

Para una call:

$\max(S - K,0)$

## Valor temporal

Es la parte del precio de la opción que refleja el **tiempo restante hasta el vencimiento**.

Incluso si una opción no tiene valor intrínseco hoy, puede tener valor porque existe la posibilidad de que el precio del activo cambie favorablemente antes del vencimiento.

A medida que se acerca la fecha de vencimiento, este valor tiende a disminuir.

Este fenómeno se conoce como **decaimiento temporal**.

# 10. Valuación de opciones como valor esperado

El valor de una opción depende del **precio del activo en el vencimiento**, ($S_T$).

Por ejemplo, el payoff de una call europea es

$\max(S_T - K,0)$

El precio teórico de la opción puede interpretarse como el **valor esperado descontado del payoff futuro**.

Bajo el marco de valuación neutral al riesgo:

$C = e^{-rT}E[\max(S_T - K,0)]$

donde

- $(r)$ es la tasa libre de riesgo
- $(T)$ es el tiempo al vencimiento
- $(E[\cdot])$ representa el valor esperado.

La interpretación es:

1. calcular el payoff futuro de la opción
2. tomar el valor esperado de ese payoff
3. descontarlo al presente

---

# 11. Idea del método de Monte Carlo

El método de **Monte Carlo** permite aproximar este valor esperado mediante simulación.

La idea es:

1. simular muchos posibles valores futuros del precio $(S_T)$
2. calcular el payoff de la opción en cada escenario
3. promediar esos payoffs
4. descontar el resultado al presente

Formalmente:

$C \approx e^{-rT}\frac{1}{N}\sum_{i=1}^{N}\max(S_T^{(i)}-K,0)$

donde:

- $(N)$ es el número de simulaciones
- $(S_T^{(i)})$ es el precio del activo en el escenario $i$

Mientras mayor sea $(N)$, mejor será la aproximación del valor esperado.
